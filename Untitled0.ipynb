```python
# ============================================================
# MINI PROJECT: STUDENT PERFORMANCE ANALYSIS
# ============================================================

students = [
    {"name": "Sumiksha", "python": 85, "sql": 82, "excel": 90},
    {"name": "Rahul", "python": 72, "sql": 78, "excel": 75},
    {"name": "Priya", "python": 92, "sql": 88, "excel": 95},
    {"name": "Aman", "python": 68, "sql": 74, "excel": 70},
    {"name": "Neha", "python": 88, "sql": 91, "excel": 86}
]

print("STUDENT PERFORMANCE ANALYSIS")
print("=" * 40)

for student in students:

    total = student["python"] + student["sql"] + student["excel"]
    average = total / 3

    student["total"] = total
    student["average"] = round(average, 2)

    print(
        student["name"],
        "| Total:", total,
        "| Average:", round(average, 2)
    )
```

```python
# Find the top-performing student

top_student = max(students, key=lambda x: x["average"])

print("\nTop Performing Student:")
print("Name:", top_student["name"])
print("Average:", top_student["average"])
```

```python
import matplotlib.pyplot as plt

names = [s["name"] for s in students]
averages = [s["average"] for s in students]

plt.figure(figsize=(8, 5))
plt.bar(names, averages)

plt.title("Student Average Performance")
plt.xlabel("Students")
plt.ylabel("Average Marks")

plt.show()
```

```python
# Calculate subject averages

python_avg = sum(s["python"] for s in students) / len(students)
sql_avg = sum(s["sql"] for s in students) / len(students)
excel_avg = sum(s["excel"] for s in students) / len(students)

print("\nSubject-wise Average")
print("Python:", round(python_avg, 2))
print("SQL:", round(sql_avg, 2))
print("Excel:", round(excel_avg, 2))
```

```python
# Subject-wise performance

subjects = ["Python", "SQL", "Excel"]
averages = [python_avg, sql_avg, excel_avg]

plt.figure(figsize=(7, 5))
plt.bar(subjects, averages)

plt.title("Subject-wise Average Performance")
plt.xlabel("Subjects")
plt.ylabel("Average Marks")

plt.show()
```
